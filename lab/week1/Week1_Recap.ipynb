## 2. print(), Comment, และตัวแปร (Variables)

```python
# นี่คือ comment บรรทัดเดียว ใช้เครื่องหมาย #
"""
นี่คือ comment หลายบรรทัด (docstring)
ใช้สามเครื่องหมายคำพูดคู่ขนาบ
"""

print("Hello, World!")
print("ผลรวมคือ:", 2 + 2)          # print() รับได้หลายค่าในครั้งเดียว
print("A", "B", "C", sep="-")      # sep กำหนดตัวคั่นระหว่างค่า -> A-B-C
print("ไม่ขึ้นบรรทัดใหม่", end=" ")  # end กำหนดสิ่งที่ต่อท้าย (ปกติคือ \n)
```

### กฎการตั้งชื่อตัวแปร
- ต้องขึ้นต้นด้วยตัวอักษรหรือ underscore (`_`) เท่านั้น
- ตัวต่อไปเป็นตัวอักษร ตัวเลข หรือ underscore ได้
- ตัวพิมพ์เล็ก/ใหญ่มีความหมายต่างกัน (case-sensitive)
- นิยมใช้ snake_case เช่น `my_age`, `user_name`
- ห้ามใช้คำสงวน (keyword) เป็นชื่อตัวแปร ดูรายการได้ด้วย:

```python
import keyword
print(keyword.kwlist)
```

```python
age = 30                # การกำหนดค่าทำงานจากขวาไปซ้ายเสมอ
0 = age                 # ❌ SyntaxError: can't assign to literal
```

### กำหนดค่าหลายตัวพร้อมกัน
```python
a, b, c = 1, 2, 3
x = y = z = 0            # x, y, z ทุกตัวอ้างถึง object ตัวเดียวกัน (เลข 0)
first, *rest = [1, 2, 3, 4]   # first = 1, rest = [2, 3, 4]
```

## 3. ประเภทข้อมูลพื้นฐาน (Built-in Data Types)

```python
age = 30                # int   -> จำนวนเต็ม
price = 19.99           # float -> ทศนิยม
name = "Alice"          # str   -> สตริง (ใช้ ' หรือ " ก็ได้)
is_student = True       # bool  -> True / False

print(type(age))        # <class 'int'>
print(isinstance(age, int))  # True
```

| ประเภท | ตัวอย่าง | ความหมาย |
|---|---|---|
| `int` | `38563846326424324` | จำนวนเต็ม ขนาดไม่จำกัด |
| `float` | `3.14`, `100.0e0` | เลขทศนิยม |
| `str` | `"Python"`, `'hello'` | ข้อความ |
| `bool` | `True`, `False` | จริง/เท็จ — จริง ๆ คือ subclass ของ int (`True == 1`) |
| `complex` | `2 + 1j` | เลขเชิงซ้อน (ใช้น้อยในงานทั่วไป) |
| `None` | `a = None` | ค่า "ไม่มีค่า" |

### Mutable vs Immutable (แก้ไขได้ vs แก้ไขไม่ได้)
```python
# Immutable: int, float, str, tuple
a = 5
a = 6        # ไม่ได้ "แก้" a เดิม แต่สร้าง object ใหม่แล้วผูกชื่อ a เข้าไปใหม่

# Mutable: list, dict, set
my_list = [1, 2, 3]
my_list[0] = 99   # แก้ไข object เดิมได้เลย -> [99, 2, 3]
```
ใช้ `id(obj)` เพื่อดู "ที่อยู่" ของ object และตรวจว่าค่าถูกแก้ไขจริง หรือถูกสร้างใหม่

## 4. การแปลงประเภทข้อมูล (Type Conversion)

```python
a = "123"
b = int(a)            # "123" -> 123

c = "123.456"
d = float(c)           # "123.456" -> 123.456
e = int(c)             # ❌ ValueError: ตัวเลขมีจุดทศนิยม แปลงตรงไม่ได้

text = "hello"
print(list(text))      # ['h', 'e', 'l', 'l', 'o']
print(set(text))        # {'h', 'e', 'l', 'o'}  (ไม่มีลำดับ ไม่ซ้ำ)
```

### สตริง: concatenation และ f-string
```python
first_name = "Alice"
last_name = "Smith"

full_name = first_name + " " + last_name        # concatenation ด้วย +
greeting = f"สวัสดี {full_name}!"                 # f-string (Python 3.6+) อ่านง่ายกว่า
print(greeting)
```

## 🛠️ โปรเจกต์เล็ก: Learning Log (สมุดบันทึกการเรียนรู้)

โปรแกรมนี้รวมทุกสิ่งที่เรียนในสัปดาห์นี้ไว้ในที่เดียว: `input()`, การแปลงประเภทข้อมูล, f-string, เงื่อนไข, และการเขียนไฟล์ — ใช้บันทึกว่าแต่ละวันเรียนอะไรไปบ้าง

บันทึกเป็นไฟล์ `learning_log.py`:

```python
from datetime import datetime

LOG_FILE = "learning_log.txt"

def add_entry():
    topic = input("วันนี้เรียนหัวข้อ Python อะไร: ")
    minutes_str = input("ใช้เวลาเรียนกี่นาที: ")
    minutes = int(minutes_str)                  # str -> int

    confident_input = input("มั่นใจในหัวข้อนี้ไหม (y/n): ").lower()
    is_confident = confident_input == "y"        # bool

    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M")
    entry = f"[{timestamp}] หัวข้อ: {topic} | เวลา: {minutes} นาที | มั่นใจ: {is_confident}\n"

    with open(LOG_FILE, "a", encoding="utf-8") as f:
        f.write(entry)

    print("\n✅ บันทึกเรียบร้อย:")
    print(entry)

def show_history():
    try:
        with open(LOG_FILE, "r", encoding="utf-8") as f:
            print("\n📒 ประวัติการเรียนทั้งหมด:")
            print(f.read())
    except FileNotFoundError:
        print("ยังไม่มีบันทึกใด ๆ เริ่มบันทึกได้เลย!")

def main():
    while True:
        print("\n--- เมนู Learning Log ---")
        print("1. บันทึกการเรียนรู้วันนี้")
        print("2. ดูประวัติทั้งหมด")
        print("3. ออกจากโปรแกรม")
        choice = input("เลือกเมนู (1-3): ")

        if choice == "1":
            add_entry()
        elif choice == "2":
            show_history()
        elif choice == "3":
            print("แล้วพบกันใหม่ สู้ๆ! 🐍")
            break
        else:
            print("กรุณาเลือก 1, 2 หรือ 3 เท่านั้น")

if __name__ == "__main__":
    main()
```

วิธีรัน:
```bash
python learning_log.py
```

โปรแกรมนี้จะสร้างไฟล์ `learning_log.txt` ขึ้นมาเรื่อย ๆ ทุกครั้งที่บันทึก — เป็นนิสัยติดตามการเรียนแบบง่าย ๆ ที่ใช้แค่ความรู้พื้นฐานสัปดาห์นี้ทั้งหมด (print, input, type conversion, f-string, if/else, loop, file I/O)
