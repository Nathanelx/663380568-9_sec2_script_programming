## 1. Control Flow คืออะไร

Control Flow คือกลไกที่ทำให้โปรแกรม **"ตัดสินใจ"** ว่าจะรันโค้ดส่วนไหน ขึ้นอยู่กับเงื่อนไขที่กำหนด — เป็นหัวใจของการเขียนโปรแกรมแบบมีตรรกะ

### Relational Operators (ตัวดำเนินการเปรียบเทียบ)
ผลลัพธ์เป็น `bool` (`True`/`False`) เสมอ

| Operator | ความหมาย |
|---|---|
| `==` | เท่ากับ |
| `!=` | ไม่เท่ากับ |
| `<` / `>` | น้อยกว่า / มากกว่า |
| `<=` / `>=` | น้อยกว่าหรือเท่ากับ / มากกว่าหรือเท่ากับ |

```python
a, b, c = 10, 20, 10
print(a == b)   # False
print(a == c)   # True
print(a <= c)   # True

s1, s2 = "python", "Python"
print(s1 == s2)              # False (ตัวพิมพ์เล็ก/ใหญ่ต่างกัน)
print(s1.lower() == s2.lower())  # True
```

### Logical Operators (ตัวดำเนินการตรรกะ)

| Operator | ความหมาย |
|---|---|
| `and` | จริงทั้งคู่ → ผลลัพธ์จริง |
| `or` | จริงอย่างน้อยหนึ่ง → ผลลัพธ์จริง |
| `not` | กลับค่าตรงข้าม |

```python
age = 25
has_license = True

eligible = (age > 21) and has_license
print(f"ได้รับส่วนลดประกัน: {eligible}")   # True

can_vote_or_drive = (age >= 18) or has_license
print(f"โหวตหรือขับรถได้: {can_vote_or_drive}")  # True
```

## 2. if / if-else / if-elif-else

```python
# if เดี่ยว
temperature = 28
if temperature > 25:
    print("วันนี้อากาศร้อนมาก!")

# if-else
number = 7
if number % 2 == 0:
    print(f"{number} เป็นเลขคู่")
else:
    print(f"{number} เป็นเลขคี่")

# if-elif-else: ระบบให้เกรด
score = float(input("กรอกคะแนนนักเรียน (0-100): "))

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
print(f"เกรดของนักเรียนคือ: {grade}")
```

> ⚠️ **สำคัญ**: โปรแกรมจะหยุดที่เงื่อนไข `elif` ตัวแรกที่เป็นจริง แล้วข้ามที่เหลือทั้งหมดทันที ลำดับการเขียนเงื่อนไขจึงสำคัญมาก

### Nested Conditionals (เงื่อนไขซ้อนเงื่อนไข)

```python
is_member = input("เป็นสมาชิกคลับหรือไม่ (yes/no): ").lower()
purchase_amount = float(input("ยอดซื้อ: $"))

if is_member == "yes":
    if purchase_amount >= 100:
        discount = purchase_amount * 0.15
    else:
        discount = purchase_amount * 0.05
else:
    if purchase_amount >= 200:
        discount = purchase_amount * 0.03
    else:
        discount = 0

print(f"ราคาสุทธิหลังหักส่วนลด: ${purchase_amount - discount:.2f}")
```

> 💡 ยิ่งซ้อนเยอะ ยิ่งอ่านยาก — ถ้าซ้อนเกิน 2-3 ชั้น ลองพิจารณาใช้ `and`/`or` รวมเงื่อนไข หรือแยกเป็นฟังก์ชันย่อยแทน

## 3. List / Dict / Set Comprehensions (รูปแบบย่อของ for-loop)

แทนที่จะเขียน for-loop ยาว ๆ เพื่อสร้างลิสต์ใหม่ ใช้ **comprehension** เขียนบรรทัดเดียวได้

```python
# วิธีเดิม (for-loop)
squares = []
for x in (1, 2, 3, 4):
    squares.append(x * x)
# squares = [1, 4, 9, 16]

# วิธีใหม่ (list comprehension) — รูปแบบ: [expression for element in iterable]
squares = [x * x for x in (1, 2, 3, 4)]
# squares = [1, 4, 9, 16]
```

### ใส่เงื่อนไขกรองข้อมูลด้วย `if`
รูปแบบ: `[expression for element in iterable if condition]`

```python
# เลขคูณของ 3 แต่ไม่ใช่คูณของ 6 ที่น้อยกว่า 50
result = [x for x in range(50) if x % 3 == 0 and x % 6 != 0]
print(result)   # [3, 9, 15, 21, 27, 33, 39, 45]

# ดึงตัวอักษรพยัญชนะที่ไม่ซ้ำจากประโยค
phrase = "Python 2026 Engine"
consonants = {c.lower() for c in phrase if c.isalpha() and c.lower() not in "aeiou"}
print(consonants)   # set comprehension -> {'p','t','h','n','2','g','...'}
```

### Dictionary Comprehension
```python
# กรองเฉพาะคู่ key-value ที่ต้องการ
scores = {"Alice": 85, "Bob": 55, "Craig": 92}
passed = {name: s for name, s in scores.items() if s >= 60}
print(passed)   # {'Alice': 85, 'Craig': 92}

# สลับ key กับ value
my_dict = {1: "a", 2: "b", 3: "c"}
swapped = {v: k for k, v in my_dict.items()}
print(swapped)  # {'a': 1, 'b': 2, 'c': 3}
```

### Generator Expression (ประหยัดหน่วยความจำ)
ใช้ `()` แทน `[]` — ข้อมูลจะถูกสร้างทีละตัว ไม่สร้างลิสต์ทั้งหมดพร้อมกัน เหมาะกับข้อมูลขนาดใหญ่

```python
total_evens_with_9 = sum(
    1 for x in range(1000)
    if x % 2 == 0 and "9" in str(x)
)
print(total_evens_with_9)   # 95
```

> ⚠️ ข้อควรระวัง: comprehension ที่ซับซ้อนเกินไปจะอ่านยาก — ถ้าอ่านแล้วงงเกินบรรทัดเดียว ให้กลับไปเขียน for-loop ปกติแทน

## 🛠️ อัปเดตโปรเจกต์: Learning Log v2 (เพิ่ม Control Flow + Comprehension)

ต่อยอดจาก `learning_log.py` ของสัปดาห์ที่แล้ว — เพิ่มฟีเจอร์ให้คะแนนความเข้าใจ (1-5) แบบมีเงื่อนไข และใช้ comprehension สรุปสถิติจากบันทึกทั้งหมด

```python
from datetime import datetime

LOG_FILE = "learning_log.txt"

def add_entry():
    topic = input("วันนี้เรียนหัวข้อ Python อะไร: ")
    minutes = int(input("ใช้เวลาเรียนกี่นาที: "))
    confidence = int(input("ให้คะแนนความมั่นใจ (1-5): "))

    # if-elif-else: แปลงคะแนนเป็นข้อความ
    if confidence >= 4:
        level = "มั่นใจมาก"
    elif confidence >= 2:
        level = "พอเข้าใจ"
    else:
        level = "ต้องทบทวนเพิ่ม"

    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M")
    entry = f"{timestamp}|{topic}|{minutes}|{confidence}|{level}\n"

    with open(LOG_FILE, "a", encoding="utf-8") as f:
        f.write(entry)

    print(f"\n✅ บันทึกแล้ว: {topic} ({level})")

def show_summary():
    """ใช้ comprehension สรุปสถิติจากไฟล์ log"""
    try:
        with open(LOG_FILE, "r", encoding="utf-8") as f:
            lines = f.readlines()
    except FileNotFoundError:
        print("ยังไม่มีบันทึก เริ่มบันทึกก่อนนะ!")
        return

    # list comprehension: แตกแต่ละบรรทัดเป็น list ของฟิลด์
    records = [line.strip().split("|") for line in lines if line.strip()]

    # list comprehension + condition: รวมเวลาทั้งหมดที่เรียน
    total_minutes = sum(int(r[2]) for r in records)

    # list comprehension: หัวข้อที่ "ต้องทบทวนเพิ่ม" เท่านั้น
    need_review = [r[1] for r in records if r[4] == "ต้องทบทวนเพิ่ม"]

    print(f"\n📊 สรุปการเรียนทั้งหมด: {len(records)} ครั้ง")
    print(f"⏱️  เวลารวม: {total_minutes} นาที")
    if need_review:
        print(f"⚠️  หัวข้อที่ควรทบทวนซ้ำ: {', '.join(need_review)}")
    else:
        print("🎉 ไม่มีหัวข้อที่ต้องทบทวนเพิ่ม เก่งมาก!")

def main():
    while True:
        print("\n--- เมนู Learning Log v2 ---")
        print("1. บันทึกการเรียนรู้วันนี้")
        print("2. ดูสรุปสถิติ (ใช้ comprehension)")
        print("3. ออกจากโปรแกรม")
        choice = input("เลือกเมนู (1-3): ")

        if choice == "1":
            add_entry()
        elif choice == "2":
            show_summary()
        elif choice == "3":
            print("แล้วพบกันใหม่ สู้ๆ! 🐍")
            break
        else:
            print("กรุณาเลือก 1, 2 หรือ 3 เท่านั้น")

if __name__ == "__main__":
    main()
```
