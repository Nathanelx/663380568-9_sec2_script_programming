```python
# ── Recap: List operations ──────────────────────────────────────────────────
fruits = ["mango", "banana", "cherry", "apple"]

# slicing
print(fruits[1:3])        # ['banana', 'cherry']
print(fruits[::-1])       # reverse

# methods
fruits.append("durian")
fruits.sort()
fruits.remove("banana")
print(fruits)

# comprehension
long_names = [f.upper() for f in fruits if len(f) > 5]
print(long_names)
```

## 📖 Part A: Dictionary — เหมือนสมุดโทรศัพท์ดิจิทัล

> **Real-world analogy**: Dict คือ "สมุดโทรศัพท์" — คุณค้นหาด้วย **ชื่อ (key)** ไม่ใช่หน้า (index)  
> ฐานข้อมูลของทุกแอปใช้ key-value เป็นพื้นฐาน: user profile, product catalog, config file


```python
# ── A1: การสร้าง Dictionary ──────────────────────────────────────────────────
# วิธีที่ 1: {} syntax
student = {
    "name":    "Alice Wonderland",
    "age":     20,
    "gpa":     3.8,
    "courses": ["Python", "Data Structures", "Calculus"],  # value เป็น list ได้
    "active":  True
}

# วิธีที่ 2: dict() constructor
config = dict(host="localhost", port=8080, debug=True)

# วิธีที่ 3: จาก 2 list ด้วย zip()  (จาก p5 Advanced Dictionaries)
fields = ("name", "age", "job")
values = ["Dev", 28, "Engineer"]
profile = dict(zip(fields, values))

print("student:", student)
print("config: ", config)
print("profile:", profile)
```

```python
# ── A2: การเข้าถึงค่า — [] vs .get() ────────────────────────────────────────
student = {"name": "Alice", "age": 20, "gpa": 3.8}

# วิธีที่ 1: [] — ถ้า key ไม่มีจะ KeyError ทันที
print(student["name"])    # Alice

# วิธีที่ 2: .get() — ปลอดภัยกว่า คืน None หรือ default ถ้าไม่เจอ key
print(student.get("age"))           # 20
print(student.get("city"))          # None — ไม่ crash
print(student.get("city", "ไม่ระบุ"))  # ไม่ระบุ — กำหนด default ได้

# ⚠️ KeyError demo — กับดักที่พบบ่อย
try:
    print(student["city"])          # ❌ KeyError!
except KeyError as e:
    print(f"KeyError: {e} — ใช้ .get() แทน!")
```

```python
# ── A3: เพิ่ม / แก้ไข / ลบ ──────────────────────────────────────────────────
student = {"name": "Alice", "age": 20}

# เพิ่ม key ใหม่
student["email"] = "alice@email.com"
student["city"]  = "Khon Kaen"
print("หลังเพิ่ม:", student)

# แก้ไขค่าที่มีอยู่แล้ว
student["age"] = 21
print("หลังแก้ age:", student)

# .update() — เพิ่ม/แก้ไขหลาย key พร้อมกัน
student.update({"gpa": 3.9, "major": "Computer Science"})
print("หลัง update:", student)

# Merge operator | (Python 3.9+) — จาก p5 Advanced
defaults = {"city": "Bangkok", "active": True}
merged = student | defaults    # student ชนะถ้า key ซ้ำ
print("merged:", merged)

# ลบ
del student["city"]
popped = student.pop("email")
print(f"pop 'email' → ได้ '{popped}', เหลือ:", student)
```

```python
# ── A4: การวน loop ผ่าน Dict ─────────────────────────────────────────────────
product = {
    "id":    "P001",
    "name":  "Python Textbook",
    "price": 350.00,
    "stock": 42
}

# วนผ่าน keys (default)
print("=== Keys ===")
for key in product:
    print(f"  {key}")

# วนผ่าน values
print("\n=== Values ===")
for val in product.values():
    print(f"  {val}")

# วนผ่าน key-value pairs — ใช้บ่อยที่สุด ✅
print("\n=== Items (key-value) ===")
for key, val in product.items():
    print(f"  {key:8s}: {val}")

# เช็กว่า key มีอยู่ไหม
print("\n'price' in product:", "price" in product)     # True
print("'color' in product:", "color" in product)       # False
```

```python
# ── A5: Nested Dictionary — dict ซ้อน dict ─────────────────────────────────
# Real-world: ฐานข้อมูลผู้ติดต่อ
contacts = {
    "Alice": {"phone": "081-234-5678", "email": "alice@mail.com", "city": "Khon Kaen"},
    "Bob":   {"phone": "089-876-5432", "email": "bob@mail.com",   "city": "Bangkok"},
}

# เข้าถึง nested value
print(contacts["Alice"]["phone"])
print(contacts["Bob"].get("city", "ไม่ทราบ"))

# วน loop ผ่าน nested dict
print("\n=== สมุดรายชื่อ ===")
for name, info in contacts.items():
    print(f"\n👤 {name}")
    for field, value in info.items():
        print(f"   {field}: {value}")

# dict comprehension — กรองเฉพาะ key ที่ต้องการ (จาก p5)
kk_contacts = {name: info for name, info in contacts.items()
               if info["city"] == "Khon Kaen"}
print("\nKhon Kaen contacts:", list(kk_contacts.keys()))
```

```python
# ── A6: Advanced — Safe Access & collections (จาก p5 Masterclass) ──────────
from collections import Counter, defaultdict

# Counter — นับความถี่ได้เลยในบรรทัดเดียว
text = "banana"
letter_count = Counter(text)
print("Counter:", letter_count)           # {'a': 3, 'n': 2, 'b': 1}
print("Most common:", letter_count.most_common(2))

# defaultdict — ไม่ต้องตรวจก่อนว่า key มีอยู่ไหม
word_groups = defaultdict(list)
words = ["apple", "ant", "banana", "bear", "cherry"]
for word in words:
    word_groups[word[0]].append(word)    # จัดกลุ่มตามตัวอักษรแรก
print("\nWord groups:", dict(word_groups))

# .setdefault() — ดึงหรือสร้าง key ถ้าไม่มี (จาก p5)
scores = {}
scores.setdefault("Alice", []).append(85)
scores.setdefault("Alice", []).append(92)   # ไม่ reset — append ต่อ
scores.setdefault("Bob",   []).append(78)
print("\nScores:", scores)
```

## 📖 Part B: Set — คอลเลกชันที่ไม่ซ้ำกัน

> **Real-world analogy**: Set คือ "รายชื่อสมาชิก" ที่ชื่อห้ามซ้ำ  
> ใช้งานจริงใน: ระบบ tag, รายชื่อ user ที่ไม่ซ้ำ, permission roles, blocked IP list


```python
# ── B1: การสร้าง Set ────────────────────────────────────────────────────────
# จาก literal
fruits = {"apple", "banana", "cherry", "apple"}  # "apple" ซ้ำ → เหลือ 1 ตัว
print("fruits:", fruits)                           # ลำดับไม่แน่นอน!

# จาก list (ลบซ้ำ)
scores = [85, 72, 91, 85, 60, 72, 78]
unique_scores = set(scores)
print("unique scores:", sorted(unique_scores))    # sort เพื่อดูง่าย

# ⚠️ สร้าง empty set ต้องใช้ set() ไม่ใช่ {}
empty_set  = set()    # ✅ set
empty_dict = {}       # ❌ นี่คือ dict ไม่ใช่ set!
print("empty_set type: ", type(empty_set))
print("empty_dict type:", type(empty_dict))
```

```python
# ── B2: เพิ่ม / ลบ สมาชิกใน Set ──────────────────────────────────────────────
members = {"Alice", "Bob", "Craig"}

# add() — เพิ่มทีละ 1
members.add("Diana")
print("after add:", members)

# update() — เพิ่มหลายตัวพร้อมกัน
members.update(["Eve", "Frank"])
print("after update:", members)

# remove() — ลบ, ถ้าไม่มีจะ KeyError
members.remove("Frank")

# discard() — ลบ, ถ้าไม่มีก็ไม่ error (ปลอดภัยกว่า)
members.discard("Ghost")   # ไม่มี "Ghost" ก็ไม่ crash
print("after remove/discard:", members)

# in operator — ค้นหาเร็วมาก O(1) เหมือน dict
print("'Alice' in members:", "Alice" in members)     # True
print("'Ghost' in members:", "Ghost" in members)     # False
```

```python
# ── B3: Set Operations — Mathematics in Python ──────────────────────────────
# Real-world: ระบบ enrollment — ใครเรียนวิชาไหนบ้าง
python_class = {"S001", "S003", "S005", "S007", "S009"}
java_class   = {"S002", "S003", "S006", "S007", "S010"}

# UNION (|) — นักเรียนทั้งหมดไม่ซ้ำ
all_students = python_class | java_class
print(f"นักเรียนทั้งหมด (union):          {len(all_students)} คน → {sorted(all_students)}")

# INTERSECTION (&) — เรียนทั้ง 2 วิชา
both = python_class & java_class
print(f"เรียนทั้ง 2 วิชา (intersection):   {sorted(both)}")

# DIFFERENCE (-) — เรียน Python แต่ไม่เรียน Java
only_python = python_class - java_class
print(f"เฉพาะ Python เท่านั้น (difference): {sorted(only_python)}")

# SYMMETRIC DIFFERENCE (^) — เรียนแค่วิชาเดียว ไม่ทั้งคู่
exclusive = python_class ^ java_class
print(f"เรียนแค่วิชาเดียว (sym_diff):     {sorted(exclusive)}")

# subset / superset
small = {"S003", "S007"}
print(f"\n{small} เป็น subset ของ python_class: {small.issubset(python_class)}")
print(f"python_class และ java_class ไม่ทับกันเลย: {python_class.isdisjoint(java_class)}")
```

```python
# ── B4: Real-world Set Use Cases ────────────────────────────────────────────
# Use case 1: ลบข้อมูลซ้ำจาก list
raw_visits = ["alice", "bob", "alice", "charlie", "bob", "alice"]
unique_visitors = list(set(raw_visits))
print(f"Unique visitors: {sorted(unique_visitors)}")
print(f"ลด {len(raw_visits)} → {len(unique_visitors)} รายการ")

# Use case 2: ระบบ permission — ตรวจสอบสิทธิ์
user_roles    = {"viewer", "editor"}
required_roles = {"editor", "publisher"}
has_access = bool(user_roles & required_roles)
print(f"\nUser roles: {user_roles}")
print(f"Required:   {required_roles}")
print(f"มีสิทธิ์เข้าใช้: {has_access}")

# Use case 3: ระบบ tag บทความ
article_tags   = {"python", "tutorial", "beginner", "data"}
search_tags    = {"python", "beginner"}
matches = article_tags & search_tags
relevance = len(matches) / len(search_tags) * 100
print(f"\nTag match: {matches}")
print(f"ความเกี่ยวข้อง: {relevance:.0f}%")
```

## 📖 Part C: Frozenset — Set ที่ล็อคไม่ให้เปลี่ยน

> **Real-world analogy**: frozenset คือ "รายชื่อที่ตีพิมพ์แล้ว" — อ่านได้ แต่แก้ไขไม่ได้  
> ใช้เมื่อ: ต้องการ set เป็น dict key / ต้องการป้องกันการแก้ไข config โดยไม่ตั้งใจ


```python
# ── C1: frozenset — สร้างและคุณสมบัติ ──────────────────────────────────────
# สร้าง frozenset
ALLOWED_ROLES   = frozenset({"admin", "editor", "viewer"})
BLOCKED_DOMAINS = frozenset(["spam.com", "hack.net", "fake.org"])

print("ALLOWED_ROLES:  ", ALLOWED_ROLES)
print("BLOCKED_DOMAINS:", BLOCKED_DOMAINS)
print("Type:", type(ALLOWED_ROLES))

# ✅ สิ่งที่ทำได้ (read operations)
print("\n'admin' in ALLOWED_ROLES:", "admin" in ALLOWED_ROLES)
print("Length:", len(ALLOWED_ROLES))

# set operations ทำได้เหมือนกัน
extra = frozenset({"superuser"})
all_roles = ALLOWED_ROLES | extra
print("Union:", all_roles, type(all_roles))  # ผลลัพธ์เป็น frozenset ด้วย
```

```python
# ── C2: frozenset เป็น dict key ได้ — set ทำไม่ได้! ────────────────────────
# Real-world: ระบบ permission โดยกลุ่มบทบาท
permission_db = {
    frozenset({"admin"}):           ["read", "write", "delete", "admin"],
    frozenset({"editor"}):          ["read", "write"],
    frozenset({"viewer"}):          ["read"],
    frozenset({"admin", "editor"}): ["read", "write", "delete"],
}

# ตรวจสอบสิทธิ์ของผู้ใช้
user_roles = frozenset({"editor"})
permissions = permission_db.get(user_roles, ["read"])
print(f"User {user_roles} สามารถ: {permissions}")

# ❌ set ทำเป็น dict key ไม่ได้ (เพราะ mutable = unhashable)
try:
    bad = {{"admin"}: "test"}
except TypeError as e:
    print(f"\n❌ TypeError: {e}")
    print("   → ใช้ frozenset แทน set เมื่อต้องการเป็น dict key!")
```

```python
# ── C3: frozenset immutability ───────────────────────────────────────────────
roles = frozenset({"admin", "editor"})

# ❌ แก้ไขไม่ได้
try:
    roles.add("viewer")
except AttributeError as e:
    print(f"❌ {e}")

try:
    roles.remove("admin")
except AttributeError as e:
    print(f"❌ {e}")

print("\n✅ frozenset ไม่มี method add/remove/discard/pop/clear")
print("✅ ใช้เมื่อต้องการ 'ล็อค' ค่าไม่ให้ใครเปลี่ยนโดยไม่ตั้งใจ")

# เปรียบเทียบ mutable vs immutable set
mutable_set   = {"a", "b", "c"}
immutable_set = frozenset({"a", "b", "c"})
print(f"\nset     hashable: ", end="")
try:    hash(mutable_set);   print("Yes")
except: print("No — ใช้เป็น dict key ไม่ได้")
print(f"frozenset hashable: ", end="")
try:    hash(immutable_set); print("Yes — ใช้เป็น dict key ได้! ✅")
except: print("No")
```

## 🗺️ เลือก Data Structure ให้ถูกต้อง

| คำถาม | คำตอบ | ใช้ |
|---|---|---|
| ต้องการลำดับ และเปลี่ยนแปลงได้? | ใช่ | `list []` |
| ต้องการลำดับ และห้ามเปลี่ยน? | ใช่ | `tuple ()` |
| ต้องการค้นหาด้วยชื่อ/key? | ใช่ | `dict {}` |
| ต้องการเก็บค่าไม่ซ้ำ เปลี่ยนได้? | ใช่ | `set {}` |
| ต้องการเก็บค่าไม่ซ้ำ ห้ามเปลี่ยน? | ใช่ | `frozenset` |

> **กฎง่ายๆ**: ถ้าต้องการค้นหาเร็ว → dict หรือ set  
> ถ้า key เป็นตำแหน่ง (0,1,2) → list/tuple | ถ้า key เป็นชื่อ → dict


```python
# ── การเปรียบเทียบ Performance: List vs Dict vs Set ──────────────────────────
import time, random

n = 100_000
data_list = list(range(n))
data_dict = {i: i for i in range(n)}
data_set  = set(range(n))
target = random.randint(0, n)

def time_search(structure, target, label):
    start = time.perf_counter()
    for _ in range(1000):
        _ = target in structure
    elapsed = (time.perf_counter() - start) * 1000
    print(f"  {label:<12}: {elapsed:.2f} ms (1,000 ครั้ง)")

print(f"ค้นหา {target} ใน {n:,} items × 1,000 รอบ:")
time_search(data_list, target, "list O(n)")
time_search(data_dict, target, "dict O(1)")
time_search(data_set,  target, "set  O(1)")
print("\n💡 dict และ set เร็วกว่า list มากเมื่อข้อมูลใหญ่ (O(1) vs O(n))")
```

```python
# ── Activity 5.1 เฉลย — Word Frequency Counter + Visualization ─────────────
from collections import Counter
import string

def word_frequency(text):
    # ลบ punctuation และแปลงเป็น lowercase
    clean = text.lower().translate(str.maketrans("", "", string.punctuation))
    words = clean.split()

    # วิธีที่ 1: for loop แบบ manual
    counts_manual = {}
    for word in words:
        counts_manual[word] = counts_manual.get(word, 0) + 1  # .get() กับ default 0

    # วิธีที่ 2: Counter (1 บรรทัด ✨)
    counts_counter = Counter(words)

    print(f"\nประโยค: '{text}'")
    print(f"คำทั้งหมด: {len(words)} คำ | unique: {len(counts_manual)} คำ\n")

    # แสดง bar chart แบบ text
    print("📊 Word Frequency Chart:")
    print("─" * 40)
    for word, count in sorted(counts_counter.items(), key=lambda x: -x[1])[:10]:
        bar = "█" * count
        print(f"  {word:<15} {bar} ({count})")

    # Top 3 คำที่พบบ่อยที่สุด
    top3 = counts_counter.most_common(3)
    print(f"\n🏆 Top 3: {top3}")
    return counts_counter

# ทดสอบ
sample = "the quick brown fox jumps over the lazy dog the fox"
result = word_frequency(sample)

# Challenge: หาคำที่ยาวที่สุด
all_words = list(result.keys())
longest = max(all_words, key=len)
print(f"\n📏 คำที่ยาวที่สุด: '{longest}' ({len(longest)} ตัวอักษร)")
```
