# Lab 4.1: Student Management System (List Operations)
โปรแกรมจัดการรายชื่อนักเรียนด้วย List

```python
student_names = []

print('--- Add Students ---')
for i in range(3):
    name = input(f'Enter student name #{i + 1}: ').strip()
    student_names.append(name)

print('Current list:', student_names)

search_name = input('Enter a name to search for: ').strip()
if search_name in student_names:
    print(f"Found '{search_name}' at index {student_names.index(search_name)}")
else:
    print(f"'{search_name}' not found in the list")

remove_name = input('Enter a name to remove: ').strip()
if remove_name in student_names:
    student_names.remove(remove_name)
    print(f"'{remove_name}' has been removed")
else:
    print(f"'{remove_name}' not found, nothing removed")

print('Updated list:', student_names)
student_names.sort()
print('Sorted list:', student_names)
print('Total students:', len(student_names))
```

## Challenge: เมนูจัดการรายชื่อนักเรียน

```python
# รัน cell นี้แยกเพื่อใช้ระบบเมนู
student_names = []

while True:
    print('\n--- Student Management Menu ---')
    print('1. Add student')
    print('2. Remove student')
    print('3. Search student')
    print('4. Sort students')
    print('5. View students')
    print('6. Exit')
    choice = input('Choose an option (1-6): ').strip()

    if choice == '1':
        name = input('Enter a student name: ').strip()
        if name:
            student_names.append(name)
            print('Student added.')
    elif choice == '2':
        name = input('Enter a student name to remove: ').strip()
        if name in student_names:
            student_names.remove(name)
            print('Student removed.')
        else:
            print('Student not found.')
    elif choice == '3':
        name = input('Enter a student name to search: ').strip()
        if name in student_names:
            print(f'Found at index {student_names.index(name)}')
        else:
            print('Student not found.')
    elif choice == '4':
        student_names.sort()
        print('Sorted list:', student_names)
    elif choice == '5':
        print('Current list:', student_names)
        print('Total students:', len(student_names))
    elif choice == '6':
        print('Goodbye!')
        break
    else:
        print('Invalid option.')
```

# Lab 4.2: Geometric Calculations (Tuples & Immutability)

```python
import math

point1 = (3, 4)
point2 = (6, 8)

print(f'Point 1: x = {point1[0]}, y = {point1[1]}')
print(f'Point 2: x = {point2[0]}, y = {point2[1]}')

# Tuple เปลี่ยนค่าในตำแหน่งเดิมไม่ได้ (immutable)
# point1[0] = 5  # TypeError: tuple object does not support item assignment

distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)
print(f'Distance between {point1} and {point2}: {distance}')
```

# Activity 4.1: List vs. Tuple

1. **อุณหภูมิรายวันในหนึ่งสัปดาห์: List** — อาจมีการแก้ไขข้อมูลที่บันทึกผิดได้
2. **ค่าสี RGB: Tuple** — เป็นค่าคงที่ 3 ค่าและไม่ควรเปลี่ยนโดยไม่ตั้งใจ
3. **สินค้าในรถเข็น: List** — เพิ่มหรือลบสินค้าได้
4. **วันในหนึ่งสัปดาห์: Tuple** — ชื่อและลำดับเป็นข้อมูลคงที่
5. **คะแนนงานของนักเรียน: List** — คะแนนอาจถูกอัปเดตหรือแก้ไขภายหลัง
6. **พิกัด GPS ของแลนด์มาร์ก: Tuple** — ละติจูดและลองจิจูดเป็นคู่ข้อมูลที่คงที่

# Activity 4.2: Interactive To-Do List

```python
todo_list = []

while True:
    print('\n--- To-Do List Menu ---')
    print('1. Add a task')
    print('2. View tasks')
    print('3. Mark task as complete')
    print('4. Exit')
    choice = input('Choose an option (1-4): ').strip()

    if choice == '1':
        task = input('Enter a new task: ').strip()
        if task:
            todo_list.append(task)
            print('Task added.')
    elif choice == '2':
        if todo_list:
            for index, task in enumerate(todo_list, start=1):
                print(f'{index}. {task}')
        else:
            print('Your to-do list is empty.')
        print('Total tasks:', len(todo_list))
    elif choice == '3':
        if not todo_list:
            print('Your to-do list is empty.')
            continue
        for index, task in enumerate(todo_list, start=1):
            print(f'{index}. {task}')
        try:
            task_number = int(input('Enter completed task number: '))
            if 1 <= task_number <= len(todo_list):
                completed_task = todo_list.pop(task_number - 1)
                print(f'Completed: {completed_task}')
            else:
                print('Invalid task number.')
        except ValueError:
            print('Please enter a number.')
    elif choice == '4':
        print('Goodbye!')
        break
    else:
        print('Invalid option.')
```
