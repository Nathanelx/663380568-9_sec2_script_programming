# Lab 4.1: ระบบจัดการรายชื่อนักศึกษา (List Operations)

```python
# สร้างลิสต์ว่างสำหรับเก็บรายชื่อนักศึกษา
student_names = []

# เพิ่มนักศึกษา 3 คนด้วย append()
print('--- ระบบจัดการรายชื่อนักศึกษา ---')
for i in range(3):
    name = input(f'กรอกชื่อนักศึกษาคนที่ {i + 1}: ').strip()
    student_names.append(name)

print(f'รายชื่อปัจจุบัน: {student_names}')

# ค้นหาชื่อด้วย in และแสดงตำแหน่งด้วย index()
search_name = input('กรอกชื่อที่ต้องการค้นหา: ').strip()
if search_name in student_names:
    print(f"พบ '{search_name}' ที่ index {student_names.index(search_name)}")
else:
    print(f"ไม่พบ '{search_name}' ในรายชื่อ")

# ตรวจสอบชื่อก่อนลบ เพื่อป้องกันข้อผิดพลาด
remove_name = input('กรอกชื่อที่ต้องการลบ: ').strip()
if remove_name in student_names:
    student_names.remove(remove_name)
    print(f"ลบ '{remove_name}' แล้ว")
else:
    print(f"ไม่พบ '{remove_name}' จึงไม่สามารถลบได้")

print(f'รายชื่อหลังลบ: {student_names}')

# เรียงรายชื่อตามตัวอักษรด้วย sort() และนับจำนวนด้วย len()
student_names.sort()
print(f'รายชื่อที่เรียงแล้ว: {student_names}')
print(f'จำนวนนักศึกษาทั้งหมด: {len(student_names)}')
```

## Challenge 4.1: เมนูจัดการรายชื่อนักศึกษา

```python
student_names = []

print('--- ระบบจัดการรายชื่อนักศึกษา ---')
for i in range(3):
    name = input(f'กรอกชื่อนักศึกษาคนที่ {i + 1}: ').strip()
    student_names.append(name)

student_names.sort()
print(f'รายชื่อที่เรียงแล้ว: {student_names}')
print(f'จำนวนนักศึกษาทั้งหมด: {len(student_names)}')

while True:
    print('\n--- Challenge 4.1: เมนูจัดการรายชื่อนักศึกษา ---')
    print('1. เพิ่มนักศึกษา')
    print('2. ลบนักศึกษา')
    print('3. ค้นหานักศึกษา')
    print('4. เรียงรายชื่อ')
    print('5. แสดงรายชื่อและจำนวน')
    print('6. ออกจากโปรแกรม')
    choice = input('เลือกเมนู (1-6): ').strip()

    if choice == '1':
        name = input('กรอกชื่อนักศึกษา: ').strip()
        if name:
            student_names.append(name)
            print('เพิ่มรายชื่อแล้ว')
    elif choice == '2':
        name = input('กรอกชื่อที่ต้องการลบ: ').strip()
        if name in student_names:
            student_names.remove(name)
            print('ลบรายชื่อแล้ว')
        else:
            print('ไม่พบรายชื่อ')
    elif choice == '3':
        name = input('กรอกชื่อที่ต้องการค้นหา: ').strip()
        if name in student_names:
            print(f'พบที่ index {student_names.index(name)}')
        else:
            print('ไม่พบรายชื่อ')
    elif choice == '4':
        student_names.sort()
        print(f'รายชื่อที่เรียงแล้ว: {student_names}')
    elif choice == '5':
        print(f'รายชื่อปัจจุบัน: {student_names}')
        print(f'จำนวนนักศึกษาทั้งหมด: {len(student_names)}')
    elif choice == '6':
        print('จบการทำงาน')
        break
    else:
        print('กรุณาเลือกเมนู 1-6')
```

# Lab 4.2: การคำนวณเรขาคณิต (Tuples & Immutability)

```python
import math

# กำหนดจุดสองจุดด้วย tuple
point1 = (3, 4)
point2 = (6, 8)

# แสดงพิกัด x และ y
print('--- การคำนวณเรขาคณิต ---')
print(f'จุดที่ 1: x = {point1[0]}, y = {point1[1]}')
print(f'จุดที่ 2: x = {point2[0]}, y = {point2[1]}')

# ทดลองแก้ไขค่าใน tuple เพื่อแสดงว่า tuple แก้ไขค่าไม่ได้
try:
    point1[0] = 5
except TypeError as error:
    print(f'เกิดข้อผิดพลาด: {error}')

# คำนวณระยะทาง: sqrt((x2 - x1)^2 + (y2 - y1)^2)
distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)
print(f'ระยะห่างระหว่าง {point1} และ {point2}: {distance}')
```

## Challenge 4.2: คำนวณระยะทางจากพิกัดที่ผู้ใช้กรอก

```python
import math

# ให้ผู้ใช้กรอกพิกัดของจุดทั้งสองเอง โดยไม่กำหนดตัวเลขล่วงหน้า
print('--- คำนวณระยะทางจากพิกัดที่ผู้ใช้กรอก ---')

x1 = float(input('กรอกค่า x ของจุดที่ 1: '))
y1 = float(input('กรอกค่า y ของจุดที่ 1: '))
x2 = float(input('กรอกค่า x ของจุดที่ 2: '))
y2 = float(input('กรอกค่า y ของจุดที่ 2: '))

point1 = (x1, y1)
point2 = (x2, y2)
distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)

print(f'ระยะห่างระหว่าง {point1} และ {point2}: {distance}')
```

# Activity 4.2: Interactive To-Do List

```python
todo_list = []

while True:
    print('\n--- เมนู To-Do List ---')
    print('1. เพิ่มงาน')
    print('2. ดูรายการงาน')
    print('3. ทำเครื่องหมายว่างานเสร็จแล้ว')
    print('4. ออกจากโปรแกรม')
    choice = input('เลือกเมนู (1-4): ').strip()

    if choice == '1':
        task = input('กรอกรายการงาน: ').strip()
        if task:
            todo_list.append(task)
            print('เพิ่มงานแล้ว')
    elif choice == '2':
        if todo_list:
            for index, task in enumerate(todo_list, start=1):
                print(f'{index}. {task}')
        else:
            print('ยังไม่มีรายการงาน')
        print(f'จำนวนงานทั้งหมด: {len(todo_list)}')
    elif choice == '3':
        if not todo_list:
            print('ยังไม่มีรายการงาน')
            continue
        for index, task in enumerate(todo_list, start=1):
            print(f'{index}. {task}')
        try:
            task_number = int(input('กรอกหมายเลขงานที่เสร็จแล้ว: '))
            if 1 <= task_number <= len(todo_list):
                completed_task = todo_list.pop(task_number - 1)
                print(f'ทำเสร็จแล้ว: {completed_task}')
            else:
                print('หมายเลขงานไม่ถูกต้อง')
        except ValueError:
            print('กรุณากรอกตัวเลข')
    elif choice == '4':
        print('จบการทำงาน')
        break
    else:
        print('กรุณาเลือกเมนู 1-4')
```
