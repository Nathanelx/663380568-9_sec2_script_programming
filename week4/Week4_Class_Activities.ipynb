## ⏱️ 1. Week 3 Recap — 5 นาที
> หัวข้อ: Control Flow — Loops, break/continue, Nested Loops, Generators


```python
# ── Recap 3.1: for loop + range() ──────────────────────────────────────────
# range(start, stop, step) — stop ไม่รวมในผลลัพธ์เสมอ
print("=== for + range ===")
total = 0
for i in range(1, 11):       # 1 ถึง 10
    total += i                # accumulator pattern
print(f"ผลรวม 1-10 = {total}")  # 55

# นับถอยหลัง
for i in range(5, 0, -1):
    print(i, end=" ")
print("🚀 Blast off!")
```

```python
# ── Recap 3.2: while loop — init / condition / update ──────────────────────
print("=== while loop ===")
count = 3
while count > 0:
    print(f"Countdown: {count}")
    count -= 1          # ❗ ลืมบรรทัดนี้ = infinite loop
print("Done!")
```

```python
# ── Recap 3.3: break vs continue ───────────────────────────────────────────
print("=== break (หยุด loop ทันที) ===")
for i in range(1, 10):
    if i == 5:
        break
    print(i, end=" ")   # 1 2 3 4

print()
print("=== continue (ข้ามแค่ iteration นี้) ===")
for i in range(1, 10):
    if i % 2 == 0:
        continue
    print(i, end=" ")   # 1 3 5 7 9
print()
```

```python
# ── Recap 3.4: nested loops — ตาราง 5x5 ──────────────────────────────────
print("=== Nested Loops: 5x5 grid ===")
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row*col:3d}", end="")
    print()
```

```python
# ── Recap 3.5: generator (advanced) ────────────────────────────────────────
def countdown_gen(n):
    while n >= 0:
        yield n
        n -= 1

print("=== Generator ===")
for val in countdown_gen(5):
    print(val, end=" ")  # 5 4 3 2 1 0
print()

# generator ประหยัด RAM — คำนวณทีละตัว ไม่สร้างทั้ง list พร้อมกัน
squares_gen = (x**2 for x in range(1_000_000))   # ใช้ RAM น้อยมาก
print("First 5 squares:", [next(squares_gen) for _ in range(5)])
```

## 📸 2. Week 4 Snapshot — 5 นาที
> หัวข้อ: Data Structures — Lists & Tuples


```python
# ── Snapshot: Lists คืออะไร ─────────────────────────────────────────────────
fruits = ["apple", "banana", "cherry", "date"]

# indexing (บวก/ลบ)
print(fruits[0])    # apple   (ตัวแรก)
print(fruits[-1])   # date    (ตัวสุดท้าย)

# slicing [start:stop:step]
print(fruits[1:3])  # ['banana', 'cherry']
print(fruits[::2])  # ['apple', 'cherry']  ทีละ 2

# เพิ่ม / แก้ไข / ลบ
fruits.append("elderberry")
fruits[1] = "blueberry"
fruits.remove("cherry")
print(fruits)
```

```python
# ── Snapshot: Tuples คืออะไร ────────────────────────────────────────────────
point = (10, 20)   # tuple — แก้ไขไม่ได้ (immutable)

print(f"x = {point[0]}, y = {point[1]}")

# แสดง TypeError เมื่อพยายามแก้ไข
try:
    point[0] = 99
except TypeError as e:
    print(f"❌ TypeError: {e}")

# เมื่อไหรใช้ tuple?
# ✅ พิกัด GPS, สี RGB, record ที่ไม่ควรเปลี่ยน
rgb_red = (255, 0, 0)
print(f"สีแดง RGB: {rgb_red}")
```

```python
# ── Snapshot: List vs Tuple ตารางเปรียบเทียบ ───────────────────────────────
print(f"{'คุณสมบัติ':<25} {'List []':<20} {'Tuple ()'}")
print("-" * 60)
props = [
    ("แก้ไขได้",         "✅ Mutable",      "❌ Immutable"),
    ("มีลำดับ",          "✅ Ordered",      "✅ Ordered"),
    ("ข้อมูลซ้ำได้",    "✅ Yes",          "✅ Yes"),
    ("ใช้เป็น dict key", "❌ No",           "✅ Yes"),
    ("ความเร็ว",         "ปกติ",            "เร็วกว่าเล็กน้อย"),
    ("ใช้เมื่อ",         "ข้อมูลที่เปลี่ยน", "ข้อมูลคงที่"),
]
for p in props:
    print(f"{p[0]:<25} {p[1]:<20} {p[2]}")
```

```python
# ── Indexing: บวก (0 จากซ้าย) และลบ (-1 จากขวา) ────────────────────────────
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
#           0        1         2         3         4
#          -5       -4        -3        -2        -1

print(fruits[0])    # apple      ตัวแรก
print(fruits[2])    # cherry     ตัวที่ 3 (index 2)
print(fruits[-1])   # elderberry ตัวสุดท้าย
print(fruits[-2])   # date       ตัวก่อนสุดท้าย

# ⚠️ IndexError ถ้า index เกินขอบเขต
try:
    print(fruits[10])
except IndexError as e:
    print(f"❌ IndexError: {e}")
```

```python
# ── Slicing: [start:stop:step] ──────────────────────────────────────────────
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

print(fruits[1:3])    # ['banana', 'cherry']    index 1,2 (ไม่รวม 3)
print(fruits[:3])     # ['apple', 'banana', 'cherry']  จากต้นถึง 2
print(fruits[2:])     # ['cherry', 'date', 'elderberry'] จาก 2 ถึงสุดท้าย
print(fruits[-3:])    # ['cherry', 'date', 'elderberry'] 3 ตัวสุดท้าย
print(fruits[::2])    # ['apple', 'cherry', 'elderberry']  ทีละ 2
print(fruits[::-1])   # reverse — อ่านจากหลังไปหน้า
```

### Part C: การแก้ไข List — Adding & Removing (10 นาที)

```python
# ── Adding: append, insert, extend ─────────────────────────────────────────
fruits = ["apple", "banana"]

# append — เพิ่มท้าย list (ใช้บ่อยที่สุด)
fruits.append("cherry")
print(f"append: {fruits}")

# insert(index, value) — แทรกที่ตำแหน่งที่กำหนด
fruits.insert(1, "avocado")
print(f"insert(1): {fruits}")

# extend — ต่อ list อื่นเข้ามา (ต่างจาก append ที่ใส่ทั้ง list เป็น 1 element)
more_fruits = ["date", "elderberry"]
fruits.extend(more_fruits)
print(f"extend: {fruits}")

# append vs extend — กับดักที่พบบ่อย!
a = [1, 2, 3]
b = [1, 2, 3]
a.append([4, 5])    # เพิ่ม list [4,5] เป็น 1 element
b.extend([4, 5])    # เพิ่ม 4 และ 5 แยกกัน
print(f"append([4,5]): {a}")   # [1, 2, 3, [4, 5]]
print(f"extend([4,5]): {b}")   # [1, 2, 3, 4, 5]
```

```python
# ── Removing: remove, pop, del, clear ───────────────────────────────────────
fruits = ["apple", "banana", "cherry", "banana", "date"]

# remove(value) — ลบค่าแรกที่พบ
fruits.remove("banana")
print(f"remove('banana'): {fruits}")   # ลบ 'banana' ตัวแรก

# pop(index) — ลบและ return ค่าที่ตำแหน่งนั้น (default = ตัวสุดท้าย)
last = fruits.pop()
print(f"pop(): ได้ '{last}', เหลือ {fruits}")

popped = fruits.pop(0)
print(f"pop(0): ได้ '{popped}', เหลือ {fruits}")

# del — ลบโดย index หรือ slice
fruits2 = ["a", "b", "c", "d", "e"]
del fruits2[1]
print(f"del [1]: {fruits2}")
del fruits2[1:3]
print(f"del [1:3]: {fruits2}")

# ⚠️ ValueError เมื่อ remove ค่าที่ไม่มีใน list
try:
    fruits.remove("mango")
except ValueError as e:
    print(f"❌ ValueError: {e}")
```

### Part D: List Methods ที่ใช้บ่อย (7 นาที)

```python
# ── len, sort, sorted, reverse, count, index ────────────────────────────────
scores = [85, 72, 91, 85, 60, 78, 91]

print(f"จำนวนคะแนน: {len(scores)}")
print(f"คะแนนสูงสุด: {max(scores)}")
print(f"คะแนนต่ำสุด: {min(scores)}")
print(f"คะแนนรวม:   {sum(scores)}")
print(f"คะแนนเฉลี่ย: {sum(scores)/len(scores):.2f}")
print(f"จำนวนครั้งที่ได้ 85: {scores.count(85)}")
print(f"ตำแหน่งของ 91 แรก: {scores.index(91)}")

# sort() แก้ in-place; sorted() สร้าง list ใหม่
scores.sort()
print(f"sort() in-place: {scores}")
scores.sort(reverse=True)
print(f"sort(reverse): {scores}")

original = [5, 2, 8, 1]
new_sorted = sorted(original)     # original ไม่เปลี่ยน
print(f"original: {original}, sorted(): {new_sorted}")

scores.reverse()
print(f"reverse(): {scores}")
```

```python
# ── การวน loop ผ่าน list ─────────────────────────────────────────────────────
fruits = ["apple", "banana", "cherry"]

# วิธี 1: for item in list (ใช้เมื่อต้องการแค่ค่า)
print("=== for item in list ===")
for fruit in fruits:
    print(f"  ผลไม้: {fruit}")

# วิธี 2: for i in range(len(list)) (ใช้เมื่อต้องการ index ด้วย)
print("\n=== for i in range(len) ===")
for i in range(len(fruits)):
    print(f"  [{i}] {fruits[i]}")

# วิธี 3: enumerate() — ดีที่สุดเมื่อต้องการทั้ง index และค่า
print("\n=== enumerate() — Pythonic way ===")
for i, fruit in enumerate(fruits):
    print(f"  [{i}] {fruit}")

# list comprehension — สร้าง list ใหม่แบบกระชับ
upper_fruits = [f.upper() for f in fruits]
print(f"\nUppercase: {upper_fruits}")
```

### Part E: Tuples — Immutable Sequences (10 นาที)

```python
# ── การสร้าง Tuple ──────────────────────────────────────────────────────────
empty_tuple = ()
coordinates = (10.5, 20.3)            # พิกัด
rgb_red     = (255, 0, 0)             # สี RGB
dimensions  = (1920, 1080)            # ความละเอียดหน้าจอ

# Single-item tuple — ต้องมี comma!
single_wrong  = ("apple")    # นี่คือ str ไม่ใช่ tuple!
single_correct = ("apple",)  # นี่คือ tuple ที่ถูกต้อง

print(type(single_wrong))    # <class 'str'>
print(type(single_correct))  # <class 'tuple'>
print(coordinates[0])        # 10.5
print(rgb_red[-1])           # 0

# tuple unpacking — เหมือน multiple assignment
x, y = coordinates
print(f"x={x}, y={y}")

r, g, b = rgb_red
print(f"Red:{r}, Green:{g}, Blue:{b}")
```

```python
# ── Immutability — แก้ไขไม่ได้ ──────────────────────────────────────────────
point = (3, 4)

# ❌ พยายามแก้ไข → TypeError
try:
    point[0] = 99
except TypeError as e:
    print(f"❌ ไม่สามารถแก้ไข tuple: {e}")

# ❌ พยายามเพิ่ม → AttributeError (ไม่มี method append)
try:
    point.append(5)
except AttributeError as e:
    print(f"❌ ไม่มี method นี้: {e}")

# ✅ ถ้าต้องการ "แก้ไข" tuple → แปลงเป็น list ก่อน แล้วแปลงกลับ
point_list = list(point)
point_list[0] = 99
new_point = tuple(point_list)
print(f"แปลงผ่าน list: {new_point}")  # (99, 4)
```

```python
# ── เมื่อไหรเลือก List vs Tuple ─────────────────────────────────────────────
# ✅ ใช้ LIST เมื่อข้อมูลเปลี่ยนแปลงได้
shopping_cart = ["กาแฟ", "ขนมปัง"]
shopping_cart.append("นม")        # เพิ่มสินค้าได้
shopping_cart.remove("กาแฟ")     # ลบสินค้าได้

# ✅ ใช้ TUPLE เมื่อข้อมูลไม่ควรเปลี่ยน
gps_landmark = (13.7563, 100.5018)    # พิกัด Grand Palace — ไม่ควรเปลี่ยน
screen_res   = (1920, 1080)            # ความละเอียดหน้าจอ
months       = ("Jan","Feb","Mar","Apr","May","Jun",
                "Jul","Aug","Sep","Oct","Nov","Dec")

# ✅ Tuple ใช้เป็น dictionary key ได้ (list ไม่ได้)
location_data = {}
location_data[(13.75, 100.50)] = "Grand Palace"   # ✅ tuple เป็น key
print(location_data)

try:
    location_data[[13.75, 100.50]] = "Test"        # ❌ list เป็น key ไม่ได้
except TypeError as e:
    print(f"❌ {e}")
```

## 🔬 7. Lab 4.2 — Geometric Calculations (Tuples)
**Objective**: ใช้ tuple แทนพิกัด 2D คำนวณระยะห่างระหว่างจุด


```python
# ── Lab 4.2 — Distance Calculator ──────────────────────────────────────────
import math

# Task 1: กำหนดจุด 2 จุดเป็น tuple
point1 = (3, 4)
point2 = (6, 8)

# Task 2: แสดงพิกัด
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

# Task 3: แสดง TypeError เมื่อพยายามแก้ไข
print("\n--- แสดง Immutability ---")
try:
    point1[0] = 99     # ❌ ไม่สามารถแก้ไข tuple ได้
except TypeError as e:
    print(f"❌ TypeError: {e}")
    print("   → tuple ออกแบบมาให้ข้อมูลคงที่ เหมาะกับพิกัด!")

# Task 4: คำนวณระยะห่าง (Euclidean Distance)
# สูตร: √((x2-x1)² + (y2-y1)²)
dx = point2[0] - point1[0]
dy = point2[1] - point1[1]
distance = math.sqrt(dx**2 + dy**2)
print(f"\nระยะห่างระหว่าง {point1} กับ {point2} = {distance:.4f} หน่วย")

# Verify: (6-3)² + (8-4)² = 9 + 16 = 25 → √25 = 5.0
print(f"ตรวจสอบ: √({dx}² + {dy}²) = √{dx**2 + dy**2} = {distance}")
```

```python
# ── Lab 4.2 Challenge: User-input Distance Calculator ───────────────────────
import math

def get_point(label):
    print(f"\nกรอกพิกัด {label}:")
    x = float(input(f"  x: "))
    y = float(input(f"  y: "))
    return (x, y)     # return เป็น tuple

def distance(p1, p2):
    return math.sqrt((p2[0]-p1[0])**2 + (p2[1]-p1[1])**2)

def midpoint(p1, p2):
    return ((p1[0]+p2[0])/2, (p1[1]+p2[1])/2)

p1 = get_point("A")
p2 = get_point("B")

print(f"\nPoint A: {p1}")
print(f"Point B: {p2}")
print(f"ระยะห่าง: {distance(p1, p2):.4f}")
print(f"จุดกึ่งกลาง: {midpoint(p1, p2)}")
```

## 💬 9. Activity 4.2 — Interactive To-Do List
**รูปแบบ**: coding challenge รายบุคคล — ใช้ list + loop + conditionals จาก Week 2-3


```python
# ── Activity 4.2: To-Do List Manager ───────────────────────────────────────
def todo_app():
    todo_list = []

    while True:
        print("\n" + "─"*30)
        print("📝 To-Do List Manager")
        print("─"*30)
        if todo_list:
            for i, task in enumerate(todo_list, 1):
                print(f"  {i}. {task}")
        else:
            print("  (ยังไม่มี task)")
        print("─"*30)
        print("  [1] เพิ่ม task")
        print("  [2] ทำเสร็จแล้ว (ลบ task)")
        print("  [3] ดูจำนวน task ที่เหลือ")
        print("  [4] ล้าง list ทั้งหมด")
        print("  [5] ออก")

        choice = input("\nเลือก: ").strip()

        if choice == "1":
            task = input("  task ใหม่: ").strip()
            if task:
                todo_list.append(task)
                print(f"  ✅ เพิ่ม '{task}' แล้ว")
            else:
                print("  ⚠️ กรุณากรอกชื่อ task")

        elif choice == "2":
            if todo_list:
                done = input("  task ที่เสร็จแล้ว: ").strip()
                if done in todo_list:
                    todo_list.remove(done)
                    print(f"  ✅ ลบ '{done}' ออกแล้ว")
                else:
                    print(f"  ❌ ไม่พบ '{done}'")
            else:
                print("  ⚠️ ไม่มี task อยู่ใน list")

        elif choice == "3":
            print(f"  📊 เหลืออีก {len(todo_list)} task")

        elif choice == "4":
            confirm = input("  ล้าง list ทั้งหมด? (y/n): ")
            if confirm.lower() == "y":
                todo_list.clear()
                print("  ✅ ล้าง list แล้ว")

        elif choice == "5":
            print(f"  ลาก่อน! เหลือ {len(todo_list)} task ที่ยังค้างอยู่ 💪")
            break

        else:
            print("  ⚠️ กรุณาเลือก 1-5")

todo_app()
```
